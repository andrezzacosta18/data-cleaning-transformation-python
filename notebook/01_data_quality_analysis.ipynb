## Missing Values

This section identifies missing values in each column by calculating both the number and percentage of missing records. This analysis helps assess the completeness of the dataset and identify variables that may require data cleaning.

```python
# Missing Values

print("Missing Values by Column\n")


df.info() #Look if we have missing values

# The total of rows is 8.000, if some column has less than this than it has missing value

# Shows the exact number of missing values per column
missing_values = df.isnull().sum()

# Shows the percentage of missing values per column
missing_percentage = df.isnull().mean() * 100


missing_summary = pd.DataFrame({
    "Missing Values": missing_values,
    "Missing Percentage (%)": missing_percentage
})

display(missing_summary)        # Exibe o DataFrame como uma tabela formatada no Jupyter Notebook.
                                # Também poderia ser usado apenas "missing_summary" na última linha da célula.


```

## Duplicate Rows

This section identifies duplicate records in the dataset. Detecting duplicate rows is important because duplicated data can affect the accuracy of statistical analyses and data-driven decisions.

```python
duplicate_rows = df.duplicated().sum()

# Display duplicate rows only if they exist.
if duplicate_rows > 0:
    duplicate_data = df[ 
        df.duplicated()
        ] #No DataFrame (df), pegue apenas as linhas que são duplicadas.
    display(duplicate_data)
else:
    print("No duplicate rows found.")
```

## Categorical Variables Analysis

```python
categorical_columns = [
    "type",
    "rating",
    "country",
    "listed_in"
]

for column in categorical_columns:
    print(f"\n{column}")
    print(df[column].value_counts())
```

## Outlier Detection (Optional)

This section identifies potential outliers in the numerical variables using the Interquartile Range (IQR) method. Outliers are observations that fall significantly below or above the typical range of the data and may indicate unusual values or data quality issues.

```python
Q1 = df["release_year"].quantile(0.25)
Q3 = df["release_year"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - (1.5 * IQR)
upper_limit = Q3 + (1.5 * IQR)

outliers = df[
    (df["release_year"] < lower_limit) |
    (df["release_year"] > upper_limit)
]

print(f"Number of potential outliers: {len(outliers)}")
outliers[["title", "release_year"]].head(10)
```

### Interpretation

The IQR method identified 719 potential outliers in the `release_year` variable.

However, these values do not necessarily represent data quality issues. Since the dataset includes both recent productions and classic movies released many decades ago, older release years are expected and reflect the natural distribution of the data rather than errors or inconsistencies.

## Distribution of Release Years

This histogram illustrates the distribution of release years in the dataset, showing how Netflix content is concentrated across different time periods.

```python


plt.figure(figsize=(10,5))               # Creates a figure with a width of 10 and height of 5 inches.

plt.hist(
    df["release_year"],                  # Uses the release_year column.
    bins=30,                             # Divides the data into 30 intervals (bins).
    edgecolor="black"                    # Adds a black border to each bar.
)

plt.title("Distribution of Release Years")   # Adds the chart title.
plt.xlabel("Release Year")                   # Labels the x-axis.
plt.ylabel("Number of Titles")               # Labels the y-axis.

plt.show()                               # Displays the chart.
```

- No duplicate rows were identified.
